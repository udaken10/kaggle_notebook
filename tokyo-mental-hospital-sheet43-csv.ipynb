```python
# This Python 3 environment comes with many helpful analytics libraries installed
# It is defined by the kaggle/python Docker image: https://github.com/kaggle/docker-python
# For example, here's several helpful packages to load

import numpy as np # linear algebra
import pandas as pd # data processing, CSV file I/O (e.g. pd.read_csv)
import glob
import os

```

```python
def extract_data():
    files = glob.glob("/kaggle/input/datasets/kenkuroiwa/630mental/**/**/*.xlsx", recursive=True)
    dfs = []
    
    for f in files:
        if "~$" in f: # Skip temporary files
            continue
        try:
            # Extract year from path (e.g., '2021/2021/...' -> '2021')
            path_parts = f.split(os.sep)
            year = "Unknown"
            for part in path_parts:
                if part.isdigit() and len(part) == 4:
                    year = part
                    break
            
            df = pd.read_excel(f, sheet_name='提出調査票43', header=None)
            
            # Add year and filename as levels in MultiIndex
            df.index = pd.MultiIndex.from_product([[year], [f], df.index], names=['year', 'filename', 'original_index'])
            dfs.append(df)
            print(f"Loaded {f} (Year: {year})")
        except Exception as e:
            print(f"Error loading {f}: {e}")
            
    if not dfs:
        print("No data found.")
        return
        
    combined_df = pd.concat(dfs)
    combined_df = combined_df.sort_index(level='year')
    print(f"Combined shape: {combined_df.shape}")
    
    combined_df.to_pickle("combined_43.pkl")
    combined_df.to_csv("combined_43.csv") # Also save as CSV for easy viewing
    print("Saved to combined_43.pkl and combined_43.csv")

if __name__ == "__main__":
    extract_data()
```
